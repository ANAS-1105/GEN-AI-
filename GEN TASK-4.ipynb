```python
from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn",
    framework="pt"
)

print("Model loaded successfully!")
```

```python
# Enter a paragraph (200–300 words)
text = """
Artificial Intelligence (AI) is transforming the way people live and work across the world.
It enables computers to perform tasks that normally require human intelligence, such as
understanding language, recognizing images, solving problems, and making decisions.
AI is widely used in healthcare, where it helps doctors diagnose diseases, recommend
treatments, and monitor patients. In education, AI-powered tools provide personalized
learning experiences, making it easier for students to understand difficult concepts.
Businesses use AI to analyze customer behavior, automate repetitive tasks, and improve
decision-making. In transportation, self-driving cars and smart traffic management systems
are reducing accidents and improving efficiency. AI also plays a major role in entertainment
through personalized recommendations on streaming platforms and social media. Despite
its many advantages, AI raises concerns about privacy, job displacement, and ethical
decision-making. As AI systems become more advanced, governments and organizations are
working to establish regulations that ensure these technologies are used responsibly and
fairly. Researchers continue to improve AI models to make them more accurate, transparent,
and reliable. In the future, AI is expected to contribute even more significantly to scientific
research, environmental protection, and everyday life. Understanding both the opportunities
and challenges of AI is essential so that society can maximize its benefits while minimizing
its risks.
"""

# Generate summary
summary = summarizer(
    text,
    max_length=80,
    min_length=30,
    do_sample=False
)

print("Generated Summary:\n")
print(summary[0]["summary_text"])

# Compare word counts
original_count = len(text.split())
summary_count = len(summary[0]["summary_text"].split())

print("\nOriginal Word Count:", original_count)
print("Summary Word Count :", summary_count)
```
